```python
#41~50
prices = {
    "apple": 100,
    "banana": 80,
    "orange": 120
}

order_text = "apple,banana,orange"
order_list = order_text.split(',')

total_price = 0
error_found = False

for item in order_list:
    if item in prices:
        total_price += prices[item]
    else:
        print(f"エラー: {item} は価格表に存在しません")
        error_found = True

if not error_found:
    display_order = " / ".join(order_list)
    print(f"注文内容: {display_order}")
    print(f"合計金額: {total_price}円")
```
